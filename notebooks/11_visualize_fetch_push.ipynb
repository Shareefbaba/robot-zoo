# Fetch Push

```python
import numpy as np, matplotlib.pyplot as plt, json, random
from pathlib import Path
DATA = Path('/home/shareef/robot_zoo/data/demos/fetch_push')
episodes = sorted(DATA.glob('episode_*.npz'))
print(f'Total episodes: {len(episodes)}')
ep0 = np.load(episodes[0])
print('Keys:', list(ep0.keys()))
print('State shape:', ep0['observation_state'].shape)
```

```python
# Episode length distribution
lengths = [len(np.load(e)['reward']) for e in episodes]
fig, ax = plt.subplots(figsize=(10,4))
ax.hist(lengths, bins=20, color='steelblue', alpha=0.8, edgecolor='black')
ax.set_title('Fetch Push — Episode Lengths')
ax.set_xlabel('Steps'); ax.set_ylabel('Count'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(DATA/'episode_lengths.png', dpi=120, bbox_inches='tight')
plt.show()
print(f'Mean: {np.mean(lengths):.1f} | Total steps: {sum(lengths):,}')
```

```python
# Push distance in X
fig, ax = plt.subplots(figsize=(12,5))
for path in random.sample(episodes, min(10, len(episodes))):
    ep = np.load(path)
    x = ep['observation_cube_pos'][:,0]
    ax.plot(x - x[0], alpha=0.6, linewidth=1.5)
ax.axhline(0.03, color='red', linestyle='--', linewidth=2, label='Success (0.03m)')
ax.set_title('Fetch Push — Cube X Displacement')
ax.set_xlabel('Timestep'); ax.set_ylabel('Displacement (m)'); ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(DATA/'push_distance.png', dpi=120, bbox_inches='tight')
plt.show()
```
