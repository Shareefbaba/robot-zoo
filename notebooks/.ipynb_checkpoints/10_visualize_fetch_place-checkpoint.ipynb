# Fetch Place Cube

```python
import numpy as np, matplotlib.pyplot as plt, json, random
from pathlib import Path
DATA = Path('/home/shareef/robot_zoo/data/demos/fetch_place')
episodes = sorted(DATA.glob('episode_*.npz'))
print(f'Total episodes: {len(episodes)}')
ep0 = np.load(episodes[0])
print('Keys:', list(ep0.keys()))
print('State shape:', ep0['observation_state'].shape)
```

```python
# Episode length distribution
lengths = [len(np.load(e)['reward']) for e in episodes]
fig, ax = plt.subplots(figsize=(10,4))
ax.hist(lengths, bins=20, color='steelblue', alpha=0.8, edgecolor='black')
ax.set_title('Fetch Place Cube — Episode Lengths')
ax.set_xlabel('Steps'); ax.set_ylabel('Count'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(DATA/'episode_lengths.png', dpi=120, bbox_inches='tight')
plt.show()
print(f'Mean: {np.mean(lengths):.1f} | Total steps: {sum(lengths):,}')
```

```python
# Pick Z then Place Z trajectory
fig, axes = plt.subplots(1,2,figsize=(14,5))
for path in random.sample(episodes, min(8, len(episodes))):
    ep = np.load(path)
    axes[0].plot(ep['observation_cube_pos'][:,2], alpha=0.6)
    axes[1].plot(ep['observation_cube_pos'][:,0], alpha=0.6)
axes[0].axhline(0.08, color='red', linestyle='--', label='Pick threshold')
axes[0].set_title('Cube Z — Pick then Place'); axes[0].legend(); axes[0].grid(alpha=0.3)
axes[1].set_title('Cube X — movement to place target'); axes[1].grid(alpha=0.3)
plt.suptitle('Fetch Place Cube', fontsize=13)
plt.tight_layout()
plt.savefig(DATA/'place_trajectory.png', dpi=120, bbox_inches='tight')
plt.show()
```
