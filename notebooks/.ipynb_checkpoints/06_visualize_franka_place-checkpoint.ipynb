# Franka Place Cube — Visualization

```python
import numpy as np, matplotlib.pyplot as plt, json, random
from pathlib import Path
DATA = Path('/home/shareef/robot_zoo/data/demos/franka_place')
episodes = sorted(DATA.glob('episode_*.npz'))
print(f'Total episodes: {len(episodes)}')
ep0 = np.load(episodes[0])
print('Keys:', list(ep0.keys()))
print('State shape:', ep0['observation_state'].shape)
```

```python
# Cube/Target Z trajectory
fig, ax = plt.subplots(figsize=(12,5))
for path in random.sample(episodes, min(10, len(episodes))):
    ep = np.load(path)
    ax.plot(ep['observation_cube_pos'][:,2], alpha=0.6, linewidth=1.5)
ax.set_title('Franka Place Cube — Z Trajectory')
ax.set_xlabel('Timestep'); ax.set_ylabel('Z (m)'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(DATA/'z_trajectory.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# Episode length distribution
lengths = [len(np.load(e)['reward']) for e in episodes]
fig, ax = plt.subplots(figsize=(8,4))
ax.hist(lengths, bins=20, color='steelblue', alpha=0.7, edgecolor='black')
ax.set_title('Franka Place Cube — Episode Lengths')
ax.set_xlabel('Steps'); ax.set_ylabel('Count'); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(DATA/'episode_lengths.png', dpi=120, bbox_inches='tight')
plt.show()
print(f'Mean length: {np.mean(lengths):.1f} | Total steps: {sum(lengths):,}')
```

```python
# Pick and place trajectory — X position shows movement
ep = np.load(episodes[0])
fig, axes = plt.subplots(1,2,figsize=(14,5))
axes[0].plot(ep['observation_cube_pos'][:,0], label='X', color='blue')
axes[0].plot(ep['observation_cube_pos'][:,1], label='Y', color='orange')
axes[0].set_title('Cube XY — shows movement to place target')
axes[0].legend(); axes[0].grid(alpha=0.3)
axes[1].plot(ep['observation_cube_pos'][:,2], color='green')
axes[1].axhline(0.12, color='red', linestyle='--', label='Pick threshold')
axes[1].set_title('Cube Z — pick then place')
axes[1].legend(); axes[1].grid(alpha=0.3)
plt.suptitle('Franka Place Cube — Pick → Move → Place', fontsize=13)
plt.tight_layout()
plt.savefig(DATA/'place_trajectory.png', dpi=120, bbox_inches='tight')
plt.show()
```
